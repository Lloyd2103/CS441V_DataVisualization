```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df_raw = pd.read_csv('abc.csv', index_col=0)
df_raw.head()
```

```python
features = {"Partner X": "Grab",
            "Partner Y": "Bee" }
```

```python
df_raw["Shipping_Partner"] = df_raw["Shipping_Partner"].map(features)
df_raw
```

```python
df_grab = df_raw[df_raw["Shipping_Partner"] == "Grab"]
df_bee = df_raw[df_raw["Shipping_Partner"] == "Bee"]
```

```python
plt.figure(figsize=(10, 6))
plt.boxplot([df_grab['Delivery_Time_Days'], df_bee['Delivery_Time_Days']], labels=['Grab', 'Bee'])
plt.title('Box Plot of Delivery Time by Shipping Partner')
plt.ylabel('Delivery Time (Days)')
plt.show()
```

```python
import seaborn as sns
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(df_raw['Delivery_Time_Days'], bins=10, alpha=0.5, kde=True)

plt.title('Histogram of Delivery Time by Shipping Partner')
plt.xlabel('Delivery Time (Days)')
plt.ylabel('Frequency')
plt.grid(True)
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.hist(df_grab['Delivery_Time_Days'], bins=10, alpha=0.5, label='Grab')
plt.hist(df_bee['Delivery_Time_Days'], bins=10, alpha=0.5, label='Bee')
plt.title('Histogram of Delivery Time by Shipping Partner')
plt.xlabel('Delivery Time (Days)')
plt.ylabel('Frequency')
plt.legend()
plt.show()
```
