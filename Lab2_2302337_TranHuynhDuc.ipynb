```python
df_indexed = df_renamed.set_index('Country')
df_indexed
```

```python
years = list(map(str, range(1980, 2014)))
years
```

```python
df_indexed['Total'] = df_indexed[years].sum(axis=1)
```

```python
df_top5 = df_indexed.sort_values(by='Total', ascending=False).drop(index=['Total','Unknown']).head(5)
df_top5
```

```python
df_top5 = df_top5[years].transpose()
df_top5
```

```python
df_top5.index = df_top5.index.map(int)
df_top5.plot(kind='area', stacked=False, figsize=(20, 10))
plt.title('Immigration to Canada from Top 5 Countries (1980 - 2013)', fontsize=20)
plt.xlabel('Years', fontsize=16)
plt.ylabel('Number of Immigrants', fontsize=16)
plt.legend(df_top5.columns, fontsize=12)
plt.show()
```

```python
df_3loc = df_indexed.loc[['Denmark','Norway','Sweden'], years].T
df_3loc
```

```python
count, bin_edges1 = np.histogram(df_3loc, 15)
df_3loc.plot(kind ='hist',
            figsize=(10, 6),
            bins=15,
            alpha=0.6,
            xticks=bin_edges1,
            color=['coral', 'darkslateblue', 'mediumseagreen']
            )
plt.title('Histogram of Immigration from Denmark, Norway, and Sweden from 1980 - 2013')
plt.ylabel('Number of Years')
plt.xlabel('Number of Immigrants')

plt.show()
```

```python
count, bin_edges = np.histogram(df_3loc, 15)
xmin = bin_edges[0] - 10   #  first bin value is 31.0, adding buffer of 10 for aesthetic purposes
xmax = bin_edges[-1] + 10  #  last bin value is 308.0, adding buffer of 10 for aesthetic purposes

# stacked Histogram
df_3loc.plot(kind='hist',
            figsize=(10, 6),
            bins=15,
            xticks=bin_edges,
            color=['coral', 'darkslateblue', 'mediumseagreen'],
            stacked=True,
            xlim=(xmin, xmax)
            )

plt.title('Histogram of Immigration from Denmark, Norway, and Sweden from 1980 - 2013')
plt.ylabel('Number of Years')
plt.xlabel('Number of Immigrants')

plt.show()
```

```python
def plot_histogram(df: pd.DataFrame, countries : list[str],quantity: int, )->None:
        data = df.loc[countries, years].T
        count, bin_edges = np.histogram(data, quantity)
        data.plot(kind='hist',
                figsize=(10, 6),
                bins=quantity,
                alpha=0.7,
                xticks=bin_edges,
                color=['coral', 'darkslateblue', 'mediumseagreen'],
                stacked=True,
                edgecolor='black'
                )
        plt.title('Histogram of Immigration from {countries}', fontsize=16)
        plt.xlabel('Number of Immigrants', fontsize=14)
        plt.ylabel('Number of Countries', fontsize=14)
        plt.show()
```

```python
plot_histogram(df_indexed, ['China', 'India', 'Philippines'], 15)
```
